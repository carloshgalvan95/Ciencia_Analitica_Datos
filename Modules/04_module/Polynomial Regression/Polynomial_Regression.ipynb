### Polynomial Regression in Python using scikit-learn

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

The formula that defines a straight line in a linear regression model is a **polynomial**, a *linear polynomial* (anything that takes the form of **$ax + b$**). let's use an example:

$$
{3x}^{4}-{7x}^{3}+{2x}^{2}+11
$$

- **Degree of a polynomial**: the highest power (exponent) in your polynomial, in this case we have a ${4}^{th}$ degree polynomial
- **Coefficient**: each number ($3,7,2,11$) in our polynomial is a coefficient; these are the parameters that are unknown and that we will try to estimate in regression models
- **Leading term**: the term with the highest power, ${3x}^{4}$, it determines the polynomial's graph behavior
- **Leading coefficient**: the coefficient of the leading term ($3$)
- **constant term**: the *y intercept*, it never changes

**Difference between linear and polynomial regression**
The polynomial regression formula (below) is very similar to the multiple linear regression formula, given that polynomial regression is a linear model used for describing non-linear relationships.

$$y = \beta_0 + \beta_1 x + \beta_2 x^2 + \ldots + \beta_n x^n$$


Linear regression is just a **first-degree** polynomial.


## Coding a polynomial regression model with a single feature

```python
x = np.arange(0, 30)
y = [3, 4, 5, 7, 10, 8, 9, 10, 10, 23, 27, 44, 50, 63, 67, 60, 62, 70, 75, 88, 81, 87, 95, 100, 108, 135, 151, 160, 169, 179]
plt.figure(figsize=(10,6))
plt.scatter(x, y)
plt.show()
```

Since we have only one feature, the following polynomial regression formula applies:

$$y = \beta_0 + \beta_1 x + \beta_2 x^2 + \ldots + \beta_n x^n$$

In this equation the number of coefficients $\beta$ is determined by the feature's highest power (aka the degree of our polynomial); not considering $\beta_0$, because it's the intercept.

Two questions immediately arise:

- How do we establish the degree of our polynomial (and thus the number of $\beta$ s)?
- How do we create $x^2$, $x^3$ or $x^n$ when originally we have $x$ as our one and only feature?

Establishing the degree of the polynomial

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2, include_bias=False)
poly
```

Creating the new features

```python
poly_features = poly.fit_transform(x.reshape(-1, 1))
poly_features
```

Creating the polynomial regression model

```python
from sklearn.linear_model import LinearRegression
poly_reg_model = LinearRegression()
poly_reg_model.fit(poly_features, y)
y_predicted = poly_reg_model.predict(poly_features)
y_predicted
```

```python
plt.figure(figsize=(10, 6))
plt.title("Your first polynomial regression", size=16)
plt.scatter(x, y)
plt.plot(x, y_predicted, c="red")
plt.show()
```

## Coding a polynomial regression model with multiple features

```python
np.random.seed(1)
x_1 = np.absolute(np.random.randn(100, 1) * 10)
x_2 = np.absolute(np.random.randn(100, 1) * 30)
y = 2*x_1**2 + 3*x_1 + 2 + np.random.randn(100, 1)*20
```

```python
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))
axes[0].scatter(x_1, y)
axes[1].scatter(x_2, y)
axes[0].set_title("x_1 plotted")
axes[1].set_title("x_2 plotted")
plt.show()
```

```python
df = pd.DataFrame({"x_1":x_1.reshape(100,), "x_2":x_2.reshape(100,), "y":y.reshape(100,)}, index=range(0,100))
df
```

Defining the training and the test data

```python
from sklearn.model_selection import train_test_split
X, y = df[["x_1", "x_2"]], df["y"]
poly = PolynomialFeatures(degree=2, include_bias=False)
poly_features = poly.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(poly_features, y, test_size=0.3, random_state=42)
```

Creating a polynomial regression model

```python
poly_reg_model = LinearRegression()
poly_reg_model.fit(X_train, y_train)
poly_reg_y_predicted = poly_reg_model.predict(X_test)
poly_reg_y_predicted
```

```python
from sklearn.metrics import mean_squared_error
poly_reg_rmse = np.sqrt(mean_squared_error(y_test, poly_reg_y_predicted))
poly_reg_rmse
```

Creating a linear regression model to compare

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
lin_reg_model = LinearRegression()
lin_reg_model.fit(X_train, y_train)
lin_reg_y_predicted = lin_reg_model.predict(X_test)
lin_reg_y_predicted
```

```python
lin_reg_rmse = np.sqrt(mean_squared_error(y_test, lin_reg_y_predicted))
lin_reg_rmse
```

## Using pipeline

```python
X, y = df[["x_1", "x_2"]], df["y"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

from sklearn.pipeline import make_pipeline

poly_reg_model = make_pipeline(PolynomialFeatures(degree=2, include_bias=False), LinearRegression())
poly_reg_model.fit(X_train, y_train)
poly_reg_y_predicted = poly_reg_model.predict(X_test)

from sklearn.metrics import mean_squared_error
poly_reg_rmse = np.sqrt(mean_squared_error(y_test, poly_reg_y_predicted))
poly_reg_rmse
```

## Finding the best degree

```python
rmses = []
degrees = np.arange(1, 10)
min_rmse, min_deg = 1e10, 0

for deg in degrees:

    poly_reg_model = make_pipeline(PolynomialFeatures(degree=deg, include_bias=False), LinearRegression())
    poly_reg_model.fit(X_train, y_train)
    poly_reg_y_predicted = poly_reg_model.predict(X_test)

    poly_reg_rmse = np.sqrt(mean_squared_error(y_test, poly_reg_y_predicted))    
    rmses.append(poly_reg_rmse)

    # Cross-validation of degree
    if min_rmse > poly_reg_rmse:
        min_rmse = poly_reg_rmse
        min_deg = deg

# Plot and present results
print('Best degree {} with RMSE {}'.format(min_deg, min_rmse))
        
plt.plot(degrees, rmses)
plt.yscale('log')
plt.xlabel('Degree')
plt.ylabel('RMSE')
```
